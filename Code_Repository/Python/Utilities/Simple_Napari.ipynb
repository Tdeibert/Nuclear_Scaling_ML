```python
import napari 
import numpy as np 
import nd2 
import pandas as pd
import tifffile as tiff
import skimage as sk
```

```python
img = nd2.imread("/home/tdeibert/data-user-tdeibert/2026.06.10_Control/2026_06_10_Control_1015.nd2")
img2 = nd2.imread("/home/tdeibert/data-user-tdeibert/2026.06.10_Control/2026_06_10_Control_1016.nd2")
print(img.shape)
print(img2.shape)
```

```python
viewer = napari.Viewer()
viewer.add_image(img, name= "image one") 
viewer.add_image(img2)
```

```python
def plane_select(stack, z, ch):
    return stack[z, ch, :, :]
 
 
plane = plane_select(img, 25, 0)
print(plane.shape)

segmentted = sk.segmentation.watershed(plane, markers=100, compactness=0.001)

viewer.add_labels(segmentted, name="segmented")
```
