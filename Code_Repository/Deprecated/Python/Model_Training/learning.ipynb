```python
def catagorize_numbers(numbers):
    """We want categorize numbers for any given series. We will use a for loop to iterate through the numbers and categorize them """
    for n in numbers:
        if n > 0:
            print("Positive")
        elif n < 0:
            print("Negative")
        else:
            print("Zero") 
catagorize_numbers([1, -2, 0]) 
```
